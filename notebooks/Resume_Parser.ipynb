```python
!pip install PyPDF2
import PyPDF2

def pdf_to_text(file_name):
    text = ""
    with open(file_name, "rb") as f:
        reader = PyPDF2.PdfReader(f)
        for page in reader.pages:
            text += page.extract_text()
    return text

resume_text = pdf_to_text(list(uploaded.keys())[0])
print(resume_text[:2000])  # show first 2000 characters
```

### PDF Text Extraction

In this step, the uploaded PDF resume is converted into plain text using the PyPDF2 library. Each page of the PDF is read and concatenated into a single text string. This unstructured text serves as input to the transformer-based NLP models.


```python
from transformers import pipeline

ner = pipeline("ner", model="dslim/bert-base-NER", grouped_entities=True)
```

### Loading Transformer-Based Named Entity Recognition Model

Here, a pretrained BERT-based Named Entity Recognition (NER) model from Hugging Face is loaded using the pipeline API. The model is capable of identifying entities such as person names and organizations from unstructured text.

```python
entities = ner(resume_text)
entities[:10]
```

### Named Entity Recognition on Resume Text

The NER model is applied to the extracted resume text to identify and classify entities. The output consists of detected entities along with their labels such as PER (person) and ORG (organization).

```python
def parse_resume(text):
    entities = ner(text)

    names = []
    orgs = []

    for e in entities:
        if e["entity_group"] == "PER":
            names.append(e["word"])
        if e["entity_group"] == "ORG":
            orgs.append(e["word"])

    return {
        "Name": " ".join(names[:2]),  # first detected name tokens
        "Organizations": list(set(orgs))
    }

parsed_output = parse_resume(resume_text)
parsed_output
```

### Resume Information Extraction

This cell processes the NER output to extract structured information. Detected person entities are combined to identify the candidate's name, and organization entities are collected to list companies or institutions mentioned in the resume. The extracted data is stored in a structured dictionary format.


```python
skills_list = ["Python", "Machine Learning", "Deep Learning", "AWS", "SQL", "Java", "C++"]

skills_found = [skill for skill in skills_list if skill.lower() in resume_text.lower()]
parsed_output["Skills"] = skills_found

parsed_output
```

```python
import json

print(json.dumps(parsed_output, indent=4))
```

### Structured JSON Output

The extracted resume information is formatted and displayed in JSON format. Structured JSON output is useful for downstream applications such as Applicant Tracking Systems (ATS) and recruitment analytics pipelines.


```python
from transformers import pipeline

summarizer = pipeline("summarization", model="sshleifer/distilbart-cnn-12-6")

summary = summarizer(resume_text, max_length=120, min_length=40)
summary[0]["summary_text"]
```

### Transformer-Based Resume Summarization

A pretrained DistilBART summarization model is used to generate a concise professional summary of the resume. This demonstrates the application of transformer-based text generation and summarization, which is part of the Unit-1 curriculum.


```python
with open("parsed_resume.json", "w") as f:
    json.dump(parsed_output, f, indent=4)

files.download("parsed_resume.json")
```
